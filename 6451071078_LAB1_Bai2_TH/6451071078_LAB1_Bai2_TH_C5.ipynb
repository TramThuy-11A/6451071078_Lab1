```python
print("Họ và tên: Nguyễn Thuỳ Trâm")
print("MSSV: 6451071078")

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Marial': ['Married', 'Married', 'Divorced', 'Single', 'Single', 'Married', 'Married', np.nan, 
               'Married', 'Divorced', 'Married', 'Married', 'Married', 'Single', 'Married', 'Married'],
    'Cons.price.idx': [93.994, 93.994, 93.994, 93.994, 93.994, 93.994, np.nan, 93.994, 
                       93.994, 93.994, 93.994, 93.918, 93.994, 93.2, 93.2, 93.2]
}

df = pd.DataFrame(data)

# 1. Vẽ biểu đồ kiểm tra giá trị thiếu trước khi xử lý
plt.figure(figsize=(6, 4))
sns.barplot(x=df.isnull().sum().index, y=df.isnull().sum().values, palette='Reds_d')
plt.title('Số lượng giá trị thiếu trước khi xử lý')
plt.ylabel('Số lượng rỗng')
plt.show()

# 2. Tiến hành điền giá trị thiếu (Imputation)
df['Marial'] = df['Marial'].fillna(df['Marial'].mode()[0])
df['Cons.price.idx'] = df['Cons.price.idx'].fillna(df['Cons.price.idx'].median())

# 3. Kiểm tra lại sau khi điền
print("Số giá trị rỗng sau khi xử lý điền khuyết:")
print(df.isnull().sum())
```
