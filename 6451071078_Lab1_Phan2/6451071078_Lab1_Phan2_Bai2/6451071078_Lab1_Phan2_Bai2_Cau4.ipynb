```python
print("Họ và tên: Nguyễn Thuỳ Trâm")
print("MSSV: 6451071078")

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import matplotlib.pyplot as plt
import os

file_name = 'Online Retail.xlsx'
if not os.path.exists(file_name):
    for root, dirs, files in os.walk('.'):
        for f in files:
            if f.lower().startswith('online retail'):
                file_name = os.path.join(root, f)
                break

df = pd.read_excel(file_name)

df['Description'] = df['Description'].astype(str).str.strip()
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
df = df[~df['InvoiceNo'].str.contains('C')]

df['Country'].value_counts().plot(kind='barh', figsize=(15,10))
plt.title('Thống kê số dòng dữ liệu theo từng quốc gia')
plt.xlabel('Số dòng dữ liệu')
plt.ylabel('Quốc gia')
plt.tight_layout()
plt.show()
```
