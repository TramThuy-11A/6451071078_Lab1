```python
print("Họ và tên: Nguyễn Thuỳ Trâm")
print("MSSV: 6451071078")

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

file_name = 'Online Retail.xlsx'
if not os.path.exists(file_name):
    for root, dirs, files in os.walk('.'):
        for f in files:
            if f.lower().startswith('online retail'):
                file_name = os.path.join(root, f)
                break

print("=== CÂU 1: ĐỌC DỮ LIỆU VÀO CHƯƠNG TRÌNH ===")
df = pd.read_excel(file_name)
df.info()

print("\n=== CÂU 2: TIỀN XỬ LÝ DỮ LIỆU ===")
df['Description'] = df['Description'].astype(str).str.strip()
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
display(df.head(5))

print("\n=== CÂU 3: XÓA CÁC HÓA ĐƠN TÍN DỤNG (CHỨA 'C') ===")
df = df[~df['InvoiceNo'].str.contains('C')]
display(df.head(5))

print("\n=== CÂU 4: THỐNG KÊ SỐ DÒNG DỮ LIỆU THEO TỪNG QUỐC GIA ===")
print("Số dòng dữ liệu của nước Pháp (France):", len(df[df['Country'] == 'France']))

print("\n=== CÂU 5: LỌC DỮ LIỆU NƯỚC PHÁP (FRANCE) VÀ GOM NHÓM ===")
basket = df[df['Country'] == "France"].groupby(['InvoiceNo', 'Description'])['Quantity']

print("\n=== CÂU 6: CHUYỂN DỮ LIỆU VỀ DẠNG HOT ENCODING ===")
basket = basket.sum().unstack().reset_index().fillna(0).set_index('InvoiceNo')
display(basket.head(5))

print("\n=== CÂU 7: CHUYỂN DỮ LIỆU SANG ONE-HOT ENCODING ===")
def encode_data(datapoint):
    if datapoint <= 0:
        return 0
    if datapoint >= 1:
        return 1

basket = basket.map(encode_data).astype(bool)
display(basket.head(5))

print("\n=== CÂU 8: XÓA CỘT 'POSTAGE' ===")
if 'POSTAGE' in basket.columns:
    basket.drop('POSTAGE', inplace=True, axis=1)
display(basket.head(5))

print("\n=== CÂU 9: TÌM TẬP PHỔ BIẾN BẰNG APRIORI (min_sup = 7%) ===")
itemsets_apriori = apriori(basket, min_support=0.07, use_colnames=True)
display(itemsets_apriori.head(10))

print("\n=== CÂU 10: TẠO LUẬT KẾT HỢP APRIORI (min_conf = 70%) ===")
rules_apriori = association_rules(itemsets_apriori, metric="confidence", min_threshold=0.7)
rules_apriori_disp = rules_apriori.copy()
rules_apriori_disp["antecedents"] = rules_apriori_disp["antecedents"].apply(lambda x: list(x)[0]).astype("str")
rules_apriori_disp["consequents"] = rules_apriori_disp["consequents"].apply(lambda x: list(x)[0]).astype("str")

for i in range(len(rules_apriori_disp)):
    print(rules_apriori_disp.loc[i, 'antecedents'], '-->', rules_apriori_disp.loc[i, 'consequents'],
          '[', rules_apriori_disp.loc[i, 'support'], ',', rules_apriori_disp.loc[i, 'confidence'], ']')

print("\n=== CÂU 11: BIỂU DIỄN ĐỘ TIN CẬY, ĐỘ HỖ TRỢ LÊN ĐỒ THỊ PHÂN TÁN ===")
plt.figure(figsize=(5, 4))
plt.title('Association Rules - France')
plt.xlabel('support')
plt.ylabel('confidence')
sns.regplot(x=rules_apriori['support'].values, y=rules_apriori['confidence'].values, fit_reg=False)
plt.show()

print("\n=== CÂU 12: TÌM TẬP PHỔ BIẾN & LUẬT BẰNG FP-GROWTH ===")
itemsets_fp = fpgrowth(basket, min_support=0.07, use_colnames=True)
display(itemsets_fp.head(10))
rules_fp = association_rules(itemsets_fp, metric="confidence", min_threshold=0.7)
print("Số lượng luật từ Apriori:", len(rules_apriori))
print("Số lượng luật từ FP-Growth:", len(rules_fp))
```
