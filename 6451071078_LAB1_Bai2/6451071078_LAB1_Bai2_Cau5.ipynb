```python
print("Họ và tên: Nguyễn Thuỳ Trâm")
print("MSSV: 6451071078")

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

for dataset in [df]:
    dataset['relatives'] = dataset['sibsp'] + dataset['parch']
    dataset.loc[dataset['relatives'] > 0, 'not_alone'] = 0
    dataset.loc[dataset['relatives'] == 0, 'not_alone'] = 1
    dataset['not_alone'] = dataset['not_alone'].astype(int)

display(df['not_alone'].value_counts())

print("\n- Số người đi một mình là 790 người.")
print("- Số người đi cùng người thân là 519 người.\n")

axes = sns.catplot(x='relatives', y='survived', data=df, aspect=2.5, kind='point')
plt.show()

print("\nSơ đồ trên cho ta thấy, những người có 1 đến 3 người thân đi cùng có tỉ lệ sống sót cao hơn. Những người có 7 đến 10 người thân đi cùng thì có tỉ lệ sống thấp nhất.")
```
