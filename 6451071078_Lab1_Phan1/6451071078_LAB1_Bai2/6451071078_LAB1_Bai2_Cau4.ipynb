```python

print("Họ và tên: Nguyễn Thuỳ Trâm")
print("MSSV: 6451071078")


import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')
FacetGrid = sns.FacetGrid(df, row='embarked', height=4.5, aspect=1.6)
FacetGrid.map(sns.pointplot, 'pclass', 'survived', 'sex', palette=None, order=None, hue_order=None)
FacetGrid.add_legend()
plt.show()

print("Từ biểu đồ trên có thể thấy được, thuộc tính Cổng lên tàu có thể liên quan đến khả năng sống sót và phụ thuộc vào giới tính:")
print("- Phụ nữ lên tàu ở cổng Q và cổng C có cơ hội sống cao hơn và ngược lại, lên tàu ở cổng S phụ nữ có tỉ lệ chết cao hơn.")
print("- Đồng nghĩa với tỉ lệ sống sót của đàn ông lên tàu ở cổng S cao hơn so với cổng Q và C.\n")

plt.figure(figsize=(6, 4))
sns.barplot(x='pclass', y='survived', data=df)
plt.show()

print("Từ biểu đồ này có thể thấy được rõ ràng tỷ lệ sống sót của loại vé 1 là cao nhất.\n")

grid = sns.FacetGrid(df, col='survived', row='pclass', height=4.2, aspect=1.6)
grid.map(plt.hist, 'age', alpha=.5, bins=20)
grid.add_legend()
plt.show()


print("Từ đồ thị trên có thể xác nhận nhận định bên trên về loại vé 1, và có thể đưa ra thêm nhận định những người có loại vé 3 thì có tỷ lệ không sống sót lớn.")
```
